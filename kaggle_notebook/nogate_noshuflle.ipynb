```python
# ==============================================================================
# KAGGLE SINGLE-CELL v4 (NO-GATE BASELINE) | TOP-1000 | 20 FEATURES | 14 FOLDS
# Pure LSTM (time) -> Transformer (cross-section)
# 3 position schemes (rank / norm[-1,1] / top10 LS) + BRAIN-style stats + PnL charts
# ==============================================================================
import os, gc, time, math, random, warnings, glob, json
from pathlib import Path
from types import SimpleNamespace
from concurrent.futures import ThreadPoolExecutor

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as TF
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

# ---------------------------- CONFIG ------------------------------------------
KAGGLE_WH  = "/kaggle/input/datasets/phmquanghiu/datar1000aaa/russell1000.duckdb"
LOCAL_WH   = "data/warehouse/russell1000.duckdb"
WAREHOUSE  = KAGGLE_WH if os.path.exists(KAGGLE_WH) else LOCAL_WH
OUT_DIR    = Path("/kaggle/working/outputs" if os.path.exists("/kaggle/working") else "runs/kaggle_top1000")
OUT_DIR.mkdir(parents=True, exist_ok=True)
CACHE_FILE = OUT_DIR / "dataset_top1000_cache.pt"

SEED, START_DATE, TEST_START, TEST_END, N_FOLDS = 42, "2000-07-01", "2010-01-01", "2024-12-31", 14
K, TTL, BUFFER_RATIO, X_WIN, Y_WIN = 1000, 60, 0.10, 20, 30
LOOKBACK, MAX_MISSING = 60, 2

FEATURE_NORM = "rankgauss"
FEATURE_CLIP = 3.0
LABEL_CLIP   = 3.0

HIDDEN_DIM, LSTM_LAYERS, TFM_LAYERS, HEADS, DROPOUT = 64, 2, 2, 4, 0.2

BATCH_DATES      = 16
MAX_EPOCHS       = 20
MAX_EPOCHS_WARM  = 8
PATIENCE         = 4
LR, WEIGHT_DECAY = 3e-4, 1e-4
WARM_LR_MULT     = 0.5
TRAIN_SUBSAMPLE  = 640
VAL_DAYS, PURGE, EMBARGO = 400, 2, 5
TIME_DECAY_HALFLIFE = 1500
N_SEEDS     = 1
MIN_STOCKS  = 20
SOFT_TAU    = 0.1
RESUME      = True   # Luu rieng duoi _nogate de khong trung voi ban co gate

EXPERIMENTS = {
    "pearson":      dict(loss="pearson",  label="zclip"),
    "pearson_rank": dict(loss="pearson",  label="rankgauss"),
    "spearman":     dict(loss="spearman", label="rank"),
    "combo":        dict(loss="combo",    label="rankgauss"),
}
RUN = ["pearson", "spearman"]

BOOK_SIZE      = 20_000_000      # chi dung de ve PnL theo $ giong BRAIN
PLOT_COST_BPS  = 2.0             # Net Sharpe tham khao (per side), thong so BRAIN la GROSS

def seed_all(seed=SEED):
    random.seed(seed); np.random.seed(seed); torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.benchmark = True

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
N_THREADS = os.cpu_count() or 4
os.environ["POLARS_MAX_THREADS"] = str(N_THREADS)
torch.set_num_threads(N_THREADS)
seed_all(SEED)
print(f"[Init] Device: {DEVICE} | CPU threads: {N_THREADS} | Warehouse: {WAREHOUSE}")

# ------------------ 1. DATASET BUILDER (cached) -------------------------------
def _gpu_rolling_stats(x, windows):
    D_, N_ = x.shape
    valid = torch.isfinite(x)
    z   = torch.where(valid, x, 0.0).to(torch.float64)
    z0  = torch.zeros((1, N_), dtype=torch.float64, device=x.device)
    c0  = torch.zeros((1, N_), dtype=torch.int32, device=x.device)
    cs1 = torch.cat([z0, z.cumsum(dim=0)], dim=0)
    cs2 = torch.cat([z0, (z * z).cumsum(dim=0)], dim=0)
    cnt = torch.cat([c0, valid.to(torch.int32).cumsum(dim=0)], dim=0)
    means, stds = {}, {}
    for n in windows:
        pad = torch.full((n - 1, N_), float("nan"), dtype=torch.float32, device=x.device)
        s1 = cs1[n:] - cs1[:-n]; s2 = cs2[n:] - cs2[:-n]
        ok = (cnt[n:] - cnt[:-n]) == n
        m  = torch.where(ok, s1 / n, float("nan")).to(torch.float32)
        var = ((s2 - (s1 * s1) / n) / max(1, n - 1)).clamp(min=0.0)
        sd = torch.where(ok, var.sqrt(), float("nan")).to(torch.float32)
        means[n] = torch.cat([pad, m], dim=0); stds[n] = torch.cat([pad, sd], dim=0)
    return means, stds, cnt

def _gpu_rolling_extrema(x, cnt, n):
    D_, N_ = x.shape
    pad = torch.full((n - 1, N_), float("nan"), dtype=torch.float32, device=x.device)
    ok  = (cnt[n:] - cnt[:-n]) == n
    x_t = x.T.unsqueeze(1)
    rmax = TF.max_pool1d(torch.nan_to_num(x_t, nan=-1e30), kernel_size=n, stride=1).squeeze(1).T
    rmin = (-TF.max_pool1d(torch.nan_to_num(-x_t, nan=-1e30), kernel_size=n, stride=1)).squeeze(1).T
    return (torch.cat([pad, torch.where(ok, rmax, float("nan"))], dim=0),
            torch.cat([pad, torch.where(ok, rmin, float("nan"))], dim=0))

def build_or_load_dataset():
    cfg_tag = f"{START_DATE}_{TEST_END}_K{K}_TTL{TTL}_L{LOOKBACK}"
    path = CACHE_FILE if CACHE_FILE.exists() else None
    if path is None:
        hits = glob.glob("/kaggle/input/**/dataset_top1000_cache.pt", recursive=True)
        path = Path(hits[0]) if hits else None
    if path is not None:
        t0 = time.time()
        cache = torch.load(path, map_location="cpu", weights_only=False)
        if cache.get("cfg_tag") == cfg_tag:
            print(f"[Cache] Loaded {path} in {time.time()-t0:.1f}s")
            return cache
        del cache

    import duckdb
    t0 = time.time()
    with duckdb.connect(WAREHOUSE, read_only=True) as con:
        con.execute(f"PRAGMA threads={N_THREADS};")
        panel = con.sql(
            f"""SELECT security_id::BIGINT AS security_id, date::DATE AS date,
                       open::FLOAT AS open, high::FLOAT AS high, low::FLOAT AS low,
                       close::FLOAT AS close, volume::FLOAT AS volume,
                       adj_close::FLOAT AS adj_close, ret_total::FLOAT AS ret_total
                FROM silver.v_prices
                WHERE date BETWEEN DATE '{START_DATE}' - INTERVAL 400 DAY AND DATE '{TEST_END}'
                ORDER BY date, security_id"""
        ).pl()
    calendar = np.sort(panel["date"].unique().to_numpy()).astype("datetime64[D]")
    all_ids  = np.sort(panel["security_id"].unique().to_numpy()).astype(np.int64)
    D_all, N_all = len(calendar), len(all_ids)
    print(f"[Build 1] {panel.height:,} rows | {N_all} stocks | {D_all} days ({time.time()-t0:.0f}s)")

    r_all = np.searchsorted(calendar, panel["date"].to_numpy().astype("datetime64[D]"))
    c_all = np.searchsorted(all_ids, panel["security_id"].to_numpy())
    raw_cols = ["open", "high", "low", "close", "volume", "adj_close", "ret_total"]
    def _pivot_col(col_name):
        arr = np.full((D_all, N_all), np.nan, dtype=np.float32)
        arr[r_all, c_all] = panel[col_name].fill_null(np.nan).to_numpy()
        return arr
    with ThreadPoolExecutor(max_workers=min(len(raw_cols), N_THREADS)) as pool:
        mats = dict(zip(raw_cols, pool.map(_pivot_col, raw_cols)))
    del panel, r_all, c_all

    close_np, vol_np = mats["close"], mats["volume"]
    has_mat = ~np.isnan(close_np)
    dv_mat  = np.nan_to_num(close_np * vol_np, nan=0.0).astype(np.float64)
    has_cs  = np.vstack([np.zeros((1, N_all), dtype=np.int32), has_mat.astype(np.int32).cumsum(axis=0)])
    dv_cs   = np.vstack([np.zeros((1, N_all), dtype=np.float64), dv_mat.cumsum(axis=0)])
    del dv_mat
    first_idx = int(np.searchsorted(calendar, np.datetime64(START_DATE, "D")))
    last_idx  = int(np.searchsorted(calendar, np.datetime64(TEST_END, "D"), side="right")) - 1
    refreshes = list(range(first_idx, last_idx + 1, TTL))
    buf_size  = int(K * BUFFER_RATIO)
    u_rows, prev_set = [], set()
    for j, idx in enumerate(refreshes):
        elig  = has_mat[idx - 1] & ((has_cs[idx] - has_cs[idx - Y_WIN]) >= X_WIN)
        cand  = all_ids[elig]
        score = dv_cs[idx, elig] - dv_cs[idx - X_WIN, elig]
        order = np.lexsort((cand, -score))
        top   = cand[order][: K + buf_size]
        kept   = [s for s in top if s in prev_set]
        kept_s = set(kept)
        fill   = [s for s in top if s not in kept_s][: K - len(kept)]
        chosen = kept_s | set(fill)
        selected = [int(s) for s in top if s in chosen]
        r_date = calendar[idx]
        until  = calendar[refreshes[j + 1]] if j + 1 < len(refreshes) else calendar[last_idx] + np.timedelta64(1, "D")
        for slot in range(K):
            sid = selected[slot] if slot < len(selected) else -1
            u_rows.append((r_date, until, slot, sid, sid == -1))
        prev_set = set(selected)
    del has_mat, has_cs, dv_cs
    active_ids = np.sort(np.unique([r[3] for r in u_rows if not r[4]])).astype(np.int64)
    act_cols   = np.searchsorted(all_ids, active_ids)
    print(f"[Build 2] universe: {len(refreshes)} periods | {len(active_ids)} active stocks")

    with torch.no_grad():
        op  = torch.from_numpy(mats["open"][:, act_cols]).to(DEVICE)
        hi  = torch.from_numpy(mats["high"][:, act_cols]).to(DEVICE)
        lo  = torch.from_numpy(mats["low"][:, act_cols]).to(DEVICE)
        cl  = torch.from_numpy(mats["close"][:, act_cols]).to(DEVICE)
        vo  = torch.from_numpy(mats["volume"][:, act_cols]).to(DEVICE)
        adj = torch.from_numpy(mats["adj_close"][:, act_cols]).to(DEVICE)
        rt  = torch.from_numpy(mats["ret_total"][:, act_cols]).to(DEVICE)
        del mats
        D_, N_ = adj.shape
        nan_row = torch.full((1, N_), float("nan"), dtype=torch.float32, device=DEVICE)
        log_ret_raw = torch.log1p(rt.clamp(min=-0.9999))
        log_ret     = log_ret_raw.clamp(-0.2231, 0.1823)
        moms = {}
        for n in (5, 10, 20, 60):
            pad_n = torch.full((n, N_), float("nan"), dtype=torch.float32, device=DEVICE)
            moms[n] = torch.cat([pad_n, adj[n:] / adj[:-n] - 1.0], dim=0)
        sma_means, _, adj_cnt = _gpu_rolling_stats(adj, (5, 10, 20, 60))
        _, vol_stds, _        = _gpu_rolling_stats(log_ret_raw, (5, 20, 60))
        dist_smas = {n: adj / sma_means[n] - 1.0 for n in (5, 10, 20, 60)}
        rmax_20, rmin_20 = _gpu_rolling_extrema(adj, adj_cnt, 20)
        dist_high_20 = adj / rmax_20 - 1.0
        dist_low_20  = adj / rmin_20 - 1.0
        hl_range   = (hi - lo) / cl
        close_open = cl / op - 1.0
        prev_cl    = torch.cat([nan_row, cl[:-1]], dim=0)
        gap        = (op / prev_cl - 1.0).clamp(-0.5, 0.5)
        diff = torch.cat([nan_row, adj[1:] - adj[:-1]], dim=0)
        up_m, _, _ = _gpu_rolling_stats(diff.clamp(min=0.0), (14,))
        dn_m, _, _ = _gpu_rolling_stats((-diff).clamp(min=0.0), (14,))
        rsi_14 = 100.0 * up_m[14] / (up_m[14] + dn_m[14] + 1e-12)
        vo_means, _, _ = _gpu_rolling_stats(vo, (20,))
        dv_means, _, _ = _gpu_rolling_stats(cl * vo, (20,))
        vol_ratio_20 = vo / (vo_means[20] + 1e-9)
        log_dollar_volume_20 = torch.log1p(dv_means[20].clamp(min=0.0))
        feat_dict = {
            "log_ret": log_ret,
            "mom_5": moms[5], "mom_10": moms[10], "mom_20": moms[20], "mom_60": moms[60],
            "vol_5": vol_stds[5], "vol_20": vol_stds[20], "vol_60": vol_stds[60],
            "dist_sma_5": dist_smas[5], "dist_sma_10": dist_smas[10],
            "dist_sma_20": dist_smas[20], "dist_sma_60": dist_smas[60],
            "dist_high_20": dist_high_20, "dist_low_20": dist_low_20,
            "hl_range": hl_range, "close_open": close_open, "gap": gap,
            "rsi_14": rsi_14, "vol_ratio_20": vol_ratio_20,
            "log_dollar_volume_20": log_dollar_volume_20,
        }
        FEATURES = list(feat_dict.keys())
        X_gpu = torch.stack(list(feat_dict.values()), dim=-1)
        X_gpu = torch.where(torch.isinf(X_gpu), float("nan"), X_gpu)
        alive_prev = ~torch.isnan(prev_cl)
        Y_gpu = torch.where(alive_prev, rt, float("nan"))
        cut = max(0, first_idx - LOOKBACK)
        X_dense  = X_gpu[cut:].cpu().numpy()
        Y_dense  = Y_gpu[cut:].cpu().numpy()
        calendar = calendar[cut:]
        del op, hi, lo, cl, vo, adj, rt, X_gpu, Y_gpu
        if DEVICE.type == "cuda":
            torch.cuda.empty_cache()

    D_, N_, NF_ = X_dense.shape
    slots = np.full((D_, K), -1, dtype=np.int32)
    for r_date, until, slot, sid, is_pad in u_rows:
        if not is_pad:
            lo_i = np.searchsorted(calendar, r_date)
            hi_i = np.searchsorted(calendar, until)
            slots[lo_i:hi_i, slot] = np.searchsorted(active_ids, sid)
    bad_day = np.isnan(X_dense).any(axis=2).astype(np.int32)
    bad_cs  = np.vstack([np.zeros((1, N_), dtype=np.int32), np.cumsum(bad_day, axis=0)])
    decision_indices = np.where((slots[:, 0] >= 0) & (np.arange(D_) >= LOOKBACK))[0]
    cache = {"cfg_tag": cfg_tag, "calendar": calendar, "active_ids": active_ids, "FEATURES": FEATURES,
             "X_dense": X_dense, "Y_dense": Y_dense, "slots": slots, "bad_cs": bad_cs,
             "decision_indices": decision_indices}
    torch.save(cache, CACHE_FILE)
    print(f"[Build 3] done in {time.time()-t0:.0f}s | cache saved ({CACHE_FILE.stat().st_size/1e6:.0f} MB)")
    return cache

# ------------------ 2. PREPROCESSING (20 FEATURES ONLY) -----------------------
@torch.no_grad()
def cs_normalize(X, mem, mode=FEATURE_NORM, chunk=128):
    D_, N_, NF_ = X.shape
    out = torch.empty((D_, N_, NF_), dtype=torch.float16)
    for s in range(0, D_, chunk):
        e = min(D_, s + chunk)
        x = torch.from_numpy(X[s:e]).to(DEVICE)
        valid = ~torch.isnan(x)
        m = torch.from_numpy(mem[s:e]).to(DEVICE)
        use = valid & m[:, :, None]
        if mode == "rankgauss":
            ref = torch.sort(x.masked_fill(~use, float("inf")), dim=1).values.permute(0, 2, 1).contiguous()
            n_ref = use.sum(dim=1).float().clamp(min=2.0)[:, :, None]
            vals = torch.nan_to_num(x, nan=0.0).permute(0, 2, 1).contiguous()
            lo_c = torch.searchsorted(ref, vals, right=False).float()
            hi_c = torch.searchsorted(ref, vals, right=True).float()
            u = 0.5 * (lo_c + hi_c) / n_ref
            lo_b = 0.5 / n_ref
            u = torch.minimum(torch.maximum(u, lo_b), 1.0 - lo_b)
            z = (math.sqrt(2.0) * torch.erfinv(2.0 * u - 1.0)).permute(0, 2, 1)
        else:
            xm = x.masked_fill(~use, float("nan"))
            med = torch.nanmedian(xm, dim=1, keepdim=True).values
            mad = torch.nanmedian((xm - med).abs(), dim=1, keepdim=True).values * 1.4826
            z = ((x - med) / mad.clamp(min=1e-8)).clamp(-FEATURE_CLIP, FEATURE_CLIP)
        z = torch.where(valid, z, torch.zeros_like(z))
        out[s:e] = z.half().cpu()
    return out

DS = None
def setup_dataset(cache):
    t0 = time.time()
    X, Y, slots = cache["X_dense"], cache["Y_dense"], cache["slots"]
    D_, N_, NF_ = X.shape
    mem = np.zeros((D_, N_), dtype=bool)
    rr = np.repeat(np.arange(D_), slots.shape[1]); cc = slots.reshape(-1); ok = cc >= 0
    mem[rr[ok], cc[ok]] = True
    Xn = cs_normalize(X, mem)
    ds = SimpleNamespace(
        calendar=cache["calendar"], active_ids=cache["active_ids"], slots=slots,
        decision_indices=cache["decision_indices"], D=D_, N=N_, n_feat=NF_,
        Xn_g=Xn.to(DEVICE), Y_g=torch.from_numpy(Y).to(DEVICE),
        slots_g=torch.from_numpy(slots).long().to(DEVICE),
        bad_cs_g=torch.from_numpy(cache["bad_cs"]).to(DEVICE),
        look_off=torch.arange(-LOOKBACK, 0, device=DEVICE),
    )
    print(f"[Setup] 20 features normalized ({FEATURE_NORM}, NO GATE) in {time.time()-t0:.1f}s | "
          f"X on {DEVICE}: {ds.Xn_g.numel()*2/1e9:.2f} GB")
    return ds

def iter_batches(indices, batch_size=BATCH_DATES, shuffle=False, sub_k=None):
    idx = np.array(indices, copy=True)
    if shuffle:
        np.random.shuffle(idx)
    for i in range(0, len(idx), batch_size):
        t_t  = torch.as_tensor(idx[i:i + batch_size], dtype=torch.long, device=DEVICE)
        cols = DS.slots_g[t_t]
        if sub_k is not None and sub_k < cols.shape[1]:
            cols = cols[:, torch.randperm(cols.shape[1], device=DEVICE)[:sub_k]]
        safe = cols.clamp(min=0); real = cols >= 0
        miss = DS.bad_cs_g[t_t[:, None], safe] - DS.bad_cs_g[(t_t - LOOKBACK)[:, None], safe]
        mask = real & (miss <= MAX_MISSING)
        t_win = t_t[:, None, None] + DS.look_off[None, None, :]
        xw = DS.Xn_g[t_win, safe[:, :, None]].float().masked_fill(~mask[:, :, None, None], 0.0)
        yv = torch.where(real, DS.Y_g[t_t[:, None], safe], float("nan"))
        ym = mask & ~torch.isnan(yv)
        yield {"x": xw, "mask": mask, "label": torch.nan_to_num(yv, nan=0.0), "label_mask": ym,
               "date_index": t_t, "cols": cols}

# ------------------ 3. MODEL (PURE LSTM -> TRANSFORMER, NO GATE) --------------
class HybridLstmTfm(nn.Module):
    def __init__(self, n_features, hidden=HIDDEN_DIM):
        super().__init__()
        self.embed = nn.Sequential(nn.Linear(n_features, hidden), nn.GELU())
        self.lstm  = nn.LSTM(hidden, hidden, num_layers=LSTM_LAYERS, batch_first=True,
                             dropout=DROPOUT if LSTM_LAYERS > 1 else 0.0)
        self.pool_proj = nn.Linear(2 * hidden, hidden)
        self.pool_ln   = nn.LayerNorm(hidden)
        layer = nn.TransformerEncoderLayer(hidden, HEADS, dim_feedforward=hidden * 2, dropout=DROPOUT,
                                           activation="gelu", batch_first=True, norm_first=True)
        self.tfm    = nn.TransformerEncoder(layer, num_layers=TFM_LAYERS, enable_nested_tensor=False)
        self.out_ln = nn.LayerNorm(hidden)
        self.head   = nn.Sequential(nn.Linear(2 * hidden, hidden), nn.GELU(), nn.Dropout(DROPOUT), nn.Linear(hidden, 1))

    def forward(self, x, mask):
        B, K_, L_, F_ = x.shape
        h_all, _ = self.lstm(self.embed(x).reshape(B * K_, L_, -1))
        last = h_all[:, -1, :]
        att  = torch.softmax((h_all * last[:, None, :]).sum(-1) / math.sqrt(h_all.shape[-1]), dim=1)
        pooled = (att[:, :, None] * h_all).sum(1)
        e = self.pool_ln(self.pool_proj(torch.cat([last, pooled], dim=-1)))
        tokens = e.reshape(B, K_, -1)
        pad = ~mask
        empty = pad.all(dim=1)
        if empty.any():
            pad = pad.clone(); pad[empty, 0] = False
        z = self.out_ln(self.tfm(tokens, src_key_padding_mask=pad))
        scores = self.head(torch.cat([tokens, z], dim=-1)).squeeze(-1)
        return scores.masked_fill(~mask, 0.0)

# ------------------ 4. LOSSES --------------------------------------------------
def masked_corr(a, b, m):
    mf = m.float(); n = mf.sum(1, keepdim=True).clamp(min=1.0)
    a = (a - (a * mf).sum(1, keepdim=True) / n) * mf
    b = (b - (b * mf).sum(1, keepdim=True) / n) * mf
    return (a * b).sum(1) / (torch.sqrt((a * a).sum(1) + 1e-12) * torch.sqrt((b * b).sum(1) + 1e-12))

def masked_rank(x, m):
    xf = x.float().masked_fill(~m, float("-inf"))
    rank = xf.argsort(dim=1).argsort(dim=1).float()
    return rank - (~m).sum(1, keepdim=True).float()

@torch.no_grad()
def make_target(y, m, mode):
    if mode == "rank":
        t = masked_rank(y, m)
    elif mode == "rankgauss":
        n = m.sum(1, keepdim=True).float().clamp(min=2.0)
        u = ((masked_rank(y, m) + 0.5) / n).clamp(1e-4, 1 - 1e-4)
        t = math.sqrt(2.0) * torch.erfinv(2.0 * u - 1.0)
    elif mode == "zclip":
        ym  = y.float().masked_fill(~m, float("nan"))
        med = torch.nanmedian(ym, dim=1, keepdim=True).values
        mad = torch.nanmedian((ym - med).abs(), dim=1, keepdim=True).values * 1.4826 + 1e-8
        t = ((y.float() - med) / mad).clamp(-LABEL_CLIP, LABEL_CLIP)
    else:
        raise ValueError(mode)
    return torch.where(m, t, torch.zeros_like(t))

def soft_rank(p, m, tau=SOFT_TAU):
    mf = m.float(); n = mf.sum(1, keepdim=True).clamp(min=2.0)
    mu = (p * mf).sum(1, keepdim=True) / n
    sd = torch.sqrt((((p - mu) * mf) ** 2).sum(1, keepdim=True) / (n - 1) + 1e-8)
    z = (p - mu) / sd
    diff = (z.unsqueeze(2) - z.unsqueeze(1)) / tau
    return (torch.sigmoid(diff) * mf.unsqueeze(1)).sum(2) - 0.5

def compute_loss(scores, y, valid, w, kind, label_mode):
    wt = w * (valid.sum(1) >= MIN_STOCKS).float()
    if float(wt.sum()) <= 0:
        return scores.sum() * 0.0
    terms = []
    if kind in ("pearson", "combo"):
        terms.append(1.0 - masked_corr(scores, make_target(y, valid, label_mode), valid))
    if kind in ("spearman", "combo"):
        terms.append(1.0 - masked_corr(soft_rank(scores, valid), make_target(y, valid, "rank"), valid))
    per_day = sum(terms) / len(terms)
    return (per_day * wt).sum() / wt.sum()

def daily_rank_ic(scores, labels, valid):
    ic = masked_corr(masked_rank(scores, valid), masked_rank(labels, valid), valid)
    return ic[valid.sum(1) >= MIN_STOCKS]

# ------------------ 5. WALK-FORWARD FOLDS --------------------------------------
def build_walk_forward_folds(decision_idx, cal):
    d_dates = cal[decision_idx]
    t0 = int(np.searchsorted(d_dates, np.datetime64(TEST_START, "D")))
    t1 = int(np.searchsorted(d_dates, np.datetime64(TEST_END, "D"), side="right"))
    folds = []
    for i, test_pos in enumerate(np.array_split(np.arange(t0, t1), N_FOLDS), 1):
        if len(test_pos) == 0:
            continue
        hist_end  = int(test_pos[0]) - PURGE - EMBARGO
        val_start = max(0, hist_end - VAL_DAYS)
        train_end = val_start - PURGE
        folds.append({"name": f"fold_{i:02d}", "train": decision_idx[:train_end],
                      "val": decision_idx[val_start:hist_end], "test": decision_idx[test_pos]})
    return folds

# ------------------ 6. TRAIN / PREDICT -----------------------------------------
USE_AMP = DEVICE.type == "cuda"

def make_day_weights(train_idx):
    w = torch.ones(DS.D, device=DEVICE)
    if TIME_DECAY_HALFLIFE:
        age = torch.as_tensor(train_idx[-1] - train_idx, dtype=torch.float32, device=DEVICE)
        w[torch.as_tensor(train_idx, dtype=torch.long, device=DEVICE)] = 0.5 ** (age / TIME_DECAY_HALFLIFE)
    return w

@torch.no_grad()
def eval_ic(model, indices):
    model.eval(); ics = []
    for b in iter_batches(indices, BATCH_DATES):
        with torch.amp.autocast("cuda", enabled=USE_AMP):
            sc = model(b["x"], b["mask"])
        ics.append(daily_rank_ic(sc.float(), b["label"], b["label_mask"]))
    ics = torch.cat(ics) if ics else torch.empty(0)
    return float(ics.mean()) if len(ics) else -1.0

def train_one(fold, cfg, seed, day_w, init_state=None):
    seed_all(seed)
    model = HybridLstmTfm(DS.n_feat).to(DEVICE)
    warm = init_state is not None
    if warm:
        model.load_state_dict(init_state)
    lr = LR * (WARM_LR_MULT if warm else 1.0)
    epochs = MAX_EPOCHS_WARM if warm else MAX_EPOCHS
    opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=WEIGHT_DECAY)
    sched = torch.optim.lr_scheduler.ReduceLROnPlateau(opt, mode="max", factor=0.5, patience=2, min_lr=1e-5)
    scaler = torch.amp.GradScaler("cuda", enabled=USE_AMP)
    best_ic, best_state, best_ep, wait, hist = -np.inf, None, 0, 0, []
    for epoch in range(1, epochs + 1):
        model.train()
        tot, nb = torch.zeros((), device=DEVICE), 0
        for b in iter_batches(fold["train"], BATCH_DATES, shuffle=True, sub_k=TRAIN_SUBSAMPLE):
            opt.zero_grad(set_to_none=True)
            with torch.amp.autocast("cuda", enabled=USE_AMP):
                scores = model(b["x"], b["mask"])
            loss = compute_loss(scores.float(), b["label"], b["label_mask"], day_w[b["date_index"]],
                                cfg["loss"], cfg["label"])
            if not torch.isfinite(loss):
                continue
            scaler.scale(loss).backward()
            scaler.unscale_(opt)
            nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            scaler.step(opt); scaler.update()
            tot += loss.detach(); nb += 1
        val_ic = eval_ic(model, fold["val"])
        hist.append(dict(epoch=epoch, train_loss=float(tot) / max(nb, 1), val_ic=val_ic,
                         lr=opt.param_groups[0]["lr"]))
        sched.step(val_ic)
        if val_ic > best_ic:
            best_ic, best_ep, wait = val_ic, epoch, 0
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
        else:
            wait += 1
            if wait >= PATIENCE:
                break
    if best_state is not None:
        model.load_state_dict(best_state)
    return model, best_ic, best_ep, best_state, hist

@torch.no_grad()
def predict(model, indices):
    model.eval(); S, M_, Y, YM, T = [], [], [], [], []
    for b in iter_batches(indices, BATCH_DATES):
        with torch.amp.autocast("cuda", enabled=USE_AMP):
            sc = model(b["x"], b["mask"])
        S.append(sc.float().cpu()); M_.append(b["mask"].cpu()); Y.append(b["label"].cpu())
        YM.append(b["label_mask"].cpu()); T.append(b["date_index"].cpu())
    return torch.cat(S), torch.cat(M_), torch.cat(Y), torch.cat(YM), torch.cat(T).numpy()

def zs(s, m):
    mf = m.float(); n = mf.sum(1, keepdim=True).clamp(min=1.0)
    mu = (s * mf).sum(1, keepdim=True) / n
    sd = torch.sqrt((((s - mu) * mf) ** 2).sum(1, keepdim=True) / n + 1e-8)
    return (s - mu) / sd * mf

def run_experiment(name, folds):
    cfg = EXPERIMENTS[name]
    print(f"\n==================== {name.upper()} [NO-GATE] (loss={cfg['loss']}, label={cfg['label']}) ====================")
    frames, prev_states = [], [None] * N_SEEDS
    for f in folds:
        path = OUT_DIR / f"preds_nogate_{name}_{f['name']}.parquet"
        if RESUME and path.exists():
            frames.append(pd.read_parquet(path)); print(f"  {f['name']} resumed from disk"); continue
        t_f = time.time()
        day_w = make_day_weights(f["train"])
        acc, val_ics, eps, hist0 = None, [], [], None
        for s in range(N_SEEDS):
            model, v_ic, ep, state, hist = train_one(f, cfg, SEED + s, day_w, prev_states[s])
            if s == 0: hist0 = hist
            prev_states[s] = state
            sc, m, y, ym, t_idx = predict(model, f["test"])
            sc_z = zs(sc, m)
            acc = sc_z if acc is None else acc + sc_z
            val_ics.append(v_ic); eps.append(ep)
            del model
            if DEVICE.type == "cuda":
                torch.cuda.empty_cache()
        acc = acc / N_SEEDS
        test_ic = float(daily_rank_ic(acc, y, ym).mean())
        mm, yy, yym = m.numpy(), y.numpy(), ym.numpy()
        rows, sl = np.nonzero(mm)
        t_rows = t_idx[rows]
        df = pd.DataFrame({
            "date": DS.calendar[t_rows],
            "security_id": DS.active_ids[DS.slots[t_rows, sl]],
            "score": acc.numpy()[rows, sl],
            "label": np.where(yym[rows, sl], yy[rows, sl], np.nan),
            "fold": f["name"],
        })
        df.to_parquet(path, index=False)
        json.dump(hist0, open(OUT_DIR / f"hist_nogate_{name}_{f['name']}.json", "w"))
        frames.append(df)
        print(f"  {f['name']} ({DS.calendar[f['test'][0]]}..{DS.calendar[f['test'][-1]]}) | "
              f"val_IC={np.mean(val_ics):+.4f} | test_IC={test_ic:+.4f} | ep={eps} | {time.time()-t_f:.0f}s")
    preds = pd.concat(frames, ignore_index=True)
    preds.to_parquet(OUT_DIR / f"predictions_nogate_{name}.parquet", index=False)
    return preds

# ------------------ 7. SCORE -> POSITION (3 CHIEN THUAT) -----------------------
# Quy uoc: long leg tong = +1, short leg tong = -1  (gross = 2, half-book = 1)
def renorm_legs(w):
    pos, neg = w.clip(lower=0.0), (-w).clip(lower=0.0)
    return pos.div(pos.sum(axis=1), axis=0).fillna(0.0) - neg.div(neg.sum(axis=1), axis=0).fillna(0.0)

def _ew(mask):
    m = mask.astype(float)
    return m.div(m.sum(axis=1), axis=0).fillna(0.0)

def w_rank(sc):
    """1) rank -> demean -> moi leg tong = 1"""
    r = sc.rank(axis=1)
    return renorm_legs(r.sub(r.mean(axis=1), axis=0).fillna(0.0))

def w_norm11(sc):
    """2) demean score, chia max|score| ve [-1,1] (giu nguyen do lon score), scale sum|w| = 2"""
    c = sc.sub(sc.mean(axis=1), axis=0)
    c = c.div(c.abs().max(axis=1) + 1e-12, axis=0)          # in [-1, 1]
    return (c.div(c.abs().sum(axis=1) + 1e-12, axis=0) * 2.0).fillna(0.0)

def w_top10(sc, q=0.10):
    """3) long top 10%, short bottom 10%, equal weight"""
    p = sc.rank(axis=1, pct=True)
    return _ew(p > 1 - q) - _ew(p <= q)

SCHEMES = {"rank": w_rank, "norm[-1,1]": w_norm11, "top10_LS": w_top10}

def prep_mats(preds):
    df = preds.dropna(subset=["score", "label"]).copy()
    df["date"] = pd.to_datetime(df["date"])
    sc = df.pivot(index="date", columns="security_id", values="score")
    rt = df.pivot(index="date", columns="security_id", values="label")
    keep = sc.notna().sum(axis=1) >= MIN_STOCKS
    sc, rt = sc.loc[keep], rt.loc[keep]
    return sc, rt.where(sc.notna()).fillna(0.0)

def run_portfolio(w, r, cost_bps=0.0):
    gross = (w * r).sum(axis=1)
    prev, r_prev, g_prev = w.shift(1, fill_value=0.0), r.shift(1, fill_value=0.0), gross.shift(1, fill_value=0.0)
    drifted = (prev * (1.0 + r_prev)).div(1.0 + g_prev, axis=0)
    traded = (w - drifted).abs().sum(axis=1)          # gross=2 convention
    return gross, gross - traded * (cost_bps / 1e4), traded

# ------------------ 8. BRAIN-STYLE STATS ---------------------------------------
def _brain_block(pnl, traded, nlong, nshort):
    sh   = float(pnl.mean() / (pnl.std() + 1e-12) * np.sqrt(252))
    ret  = float(pnl.mean() * 252)                              # PnL / half-book
    turn = float(0.5 * traded.mean())                           # traded / gross-book
    fit  = sh * math.sqrt(abs(ret) / max(turn, 0.125))
    cum  = pnl.cumsum()
    dd   = float((np.maximum(cum.cummax(), 0.0) - cum).max())   # / half-book
    mar  = float(pnl.sum() / (traded.sum() + 1e-12) * 1e4)      # bps per $ traded
    return {"Sharpe": sh, "Turnover(%)": turn * 100, "Fitness": fit, "Returns(%)": ret * 100,
            "Drawdown(%)": dd * 100, "Margin(bps)": mar,
            "LongCount": float(nlong.mean()), "ShortCount": float(nshort.mean())}

def brain_stats(w, r):
    pnl, _, traded = run_portfolio(w, r, 0.0)
    nl, ns = (w > 1e-12).sum(axis=1), (w < -1e-12).sum(axis=1)
    agg = _brain_block(pnl, traded, nl, ns)
    net = pnl - traded * PLOT_COST_BPS / 1e4
    agg[f"NetSharpe@{PLOT_COST_BPS:g}bps"] = float(net.mean() / (net.std() + 1e-12) * np.sqrt(252))
    yearly = pd.DataFrame({y: _brain_block(pnl[pnl.index.year == y], traded[pnl.index.year == y],
                                           nl[pnl.index.year == y], ns[pnl.index.year == y])
                           for y in sorted(set(pnl.index.year))}).T
    yearly.index.name = "Year"
    return agg, yearly, pnl, traded

def plot_pnl(name, res):
    fig, ax = plt.subplots(1, 4, figsize=(30, 6))
    cmap = plt.get_cmap("tab10")
    for i, (sn, d) in enumerate(res.items()):
        cum = d["pnl"].cumsum() * BOOK_SIZE / 2 / 1e3          # $K tren half-book
        a = d["agg"]
        ax[i].plot(cum.index, cum.values, color=cmap(i + 2), lw=1.5)
        ax[i].axhline(0, color="k", lw=0.8); ax[i].grid(True, alpha=0.3)
        ax[i].set_title(f"{name} (no-gate) | {sn}\nSharpe {a['Sharpe']:.2f} | Turn {a['Turnover(%)']:.1f}% | Fit {a['Fitness']:.2f} | "
                        f"Ret {a['Returns(%)']:.2f}% | DD {a['Drawdown(%)']:.2f}% | Margin {a['Margin(bps)']:.2f}bps", fontsize=10)
        ax[i].set_ylabel("PnL ($K, book=%.0fM)" % (BOOK_SIZE / 1e6))
        ax[3].plot(cum.index, cum.values, color=cmap(i + 2), lw=1.4, label=f"{sn} (Sharpe {a['Sharpe']:.2f})")
    ax[3].axhline(0, color="k", lw=0.8); ax[3].grid(True, alpha=0.3); ax[3].legend()
    ax[3].set_title(f"{name} (no-gate) | 3 schemes overlay (gross PnL)")
    plt.tight_layout(); plt.savefig(OUT_DIR / f"pnl_3schemes_nogate_{name}.png", dpi=150); plt.show()

def plot_training(name, folds, preds):
    hs = {}
    for f in folds:
        p = OUT_DIR / f"hist_nogate_{name}_{f['name']}.json"
        if p.exists(): hs[f["name"]] = json.load(open(p))
    if not hs:
        print(f"[plot_training] no history for {name} (folds resumed from disk have none)"); return
    cmap = plt.get_cmap("viridis", len(folds))
    fig, ax = plt.subplots(2, 2, figsize=(16, 10))
    for i, (fn, h) in enumerate(hs.items()):
        e = [r["epoch"] for r in h]
        ax[0, 0].plot(e, [r["train_loss"] for r in h], color=cmap(i), lw=1.3, label=fn)
        ax[0, 1].plot(e, [r["val_ic"] for r in h], color=cmap(i), lw=1.3, label=fn)
        b = int(np.argmax([r["val_ic"] for r in h]))
        ax[0, 1].scatter(e[b], h[b]["val_ic"], color=cmap(i), s=35, zorder=3)
    ax[0, 0].set_title(f"{name} (no-gate): train loss / epoch"); ax[0, 1].set_title("Val Rank IC / epoch (dot = best)")
    ax[0, 0].legend(fontsize=7, ncol=2)
    d = preds.dropna(subset=["score", "label"])
    t_ic = d.groupby(["fold", "date"]).apply(lambda g: g["score"].rank().corr(g["label"].rank())).groupby("fold").mean()
    names = list(hs.keys()); x = np.arange(len(names))
    ax[1, 0].bar(x - 0.2, [max(r["val_ic"] for r in hs[n]) for n in names], 0.4, label="best val IC")
    ax[1, 0].bar(x + 0.2, [t_ic.get(n, np.nan) for n in names], 0.4, label="test IC")
    ax[1, 0].set_xticks(x); ax[1, 0].set_xticklabels([n[-2:] for n in names]); ax[1, 0].legend()
    ax[1, 0].set_title("Val IC vs Test IC by fold")
    ax[1, 1].bar(x - 0.2, [int(np.argmax([r["val_ic"] for r in hs[n]])) + 1 for n in names], 0.4, label="best epoch")
    ax[1, 1].bar(x + 0.2, [len(hs[n]) for n in names], 0.4, label="epochs run")
    ax[1, 1].set_xticks(x); ax[1, 1].set_xticklabels([n[-2:] for n in names]); ax[1, 1].legend()
    ax[1, 1].set_title("Early stopping")
    for a in ax.ravel(): a.grid(True, alpha=0.3)
    plt.tight_layout(); plt.savefig(OUT_DIR / f"training_nogate_{name}.png", dpi=150); plt.show()

# ------------------ 9. MAIN ----------------------------------------------------
def main():
    global DS
    cache = build_or_load_dataset()
    DS = setup_dataset(cache)
    del cache; gc.collect()

    folds = build_walk_forward_folds(DS.decision_indices, DS.calendar)
    print(f"[Folds] {len(folds)} folds | {DS.calendar[folds[0]['test'][0]]} -> {DS.calendar[folds[-1]['test'][-1]]}")

    agg_all = {}
    for name in RUN:
        preds = run_experiment(name, folds)
        plot_training(name, folds, preds)

        sc, r = prep_mats(preds)
        ric = sc.rank(axis=1).corrwith(r.where(sc.notna()).rank(axis=1), axis=1).dropna()
        print(f"\n[{name} | NO-GATE] Mean Rank IC={ric.mean():.4f} | ICIR(ann)={ric.mean()/(ric.std()+1e-9)*np.sqrt(252):.2f} "
              f"| hit={100*(ric>0).mean():.1f}%")

        res = {}
        for sn, fn in SCHEMES.items():
            w = fn(sc).reindex_like(sc).fillna(0.0)
            agg, yearly, pnl, traded = brain_stats(w, r)
            res[sn] = dict(agg=agg, yearly=yearly, pnl=pnl)
            agg_all[(name, sn)] = agg
            yearly.to_csv(OUT_DIR / f"brain_yearly_nogate_{name}_{sn.replace('[','').replace(']','').replace(',','_')}.csv")
            print(f"\n=========== {name} (NO-GATE) | {sn} ===========")
            print("Aggregate:", {k: round(v, 3) for k, v in agg.items()})
            print(yearly.round(2).to_string())
        plot_pnl(name, res)

    summary = pd.DataFrame(agg_all).T
    summary.index.names = ["experiment", "scheme"]
    print("\n================ SUMMARY (NO-GATE, gross, BRAIN-style) ================")
    print(summary.round(3).to_string())
    summary.to_csv(OUT_DIR / "brain_summary_nogate_all.csv")
    print(f"\n[Done] outputs in {OUT_DIR}")

main()
```
